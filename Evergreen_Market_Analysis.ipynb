```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('retail_store_inventory.csv')
data.head()
```

```python
data.columns
```

```python
plt.savefig('images/inventory_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data['Inventory_Level'], bins=10, kde=True)
plt.title('Distribution of Inventory Levels')
plt.xlabel('Inventory Level')
plt.ylabel('Frequency')
plt.savefig('images/inventory_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='Inventory_Level', y='Units_Sold', data=data)
plt.title('Units Sold vs Inventory Level')
plt.xlabel('Inventory Level')
plt.ylabel('Units Sold')
plt.savefig('images/units_vs_inventory.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='Price', y='Competitor_Price', data=data)
plt.title('Price vs Competitor Price')
plt.xlabel('Store Price')
plt.ylabel('Competitor Price')
plt.savefig('images/price_vs_competitor.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Category', y='Inventory_Level', data=data)
plt.title('Inventory Level by Category')
plt.xlabel('Category')
plt.ylabel('Inventory Level')
plt.xticks(rotation=45)
plt.savefig('images/category_inventory.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(data.select_dtypes(include='number').corr(), annot=True, cmap='Blues')
plt.title('Correlation Heatmap')
plt.savefig('images/correlation_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
```
